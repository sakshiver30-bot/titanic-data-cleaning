```python
import pandas as pd

df = pd.read_csv("train.csv")
print("Shape before cleaning:", df.shape)
df.info()
df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nUnique values:")
print("Sex:", df['Sex'].unique())
print("Embarked:", df['Embarked'].unique())
print("Pclass:", df['Pclass'].unique())

print("\nDuplicate PassengerIds:", df['PassengerId'].duplicated().sum())
```

```python
df_clean = df.copy()

# 1. Remove duplicates
df_clean = df_clean.drop_duplicates()

# 2. Fill missing Age with the median
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())

# 3. Fill missing Embarked with the most common value
df_clean['Embarked'] = df_clean['Embarked'].fillna(df_clean['Embarked'].mode()[0])

# 4. Cabin is ~77% missing, so drop it
df_clean = df_clean.drop(columns=['Cabin'])

# 5. Standardize text
df_clean['Sex'] = df_clean['Sex'].str.strip().str.lower()
df_clean['Embarked'] = df_clean['Embarked'].str.strip().str.upper()
df_clean['Name'] = df_clean['Name'].str.strip()

# 6. Fix data types
df_clean['Age'] = df_clean['Age'].round().astype(int)
df_clean['Pclass'] = df_clean['Pclass'].astype('category')
df_clean['Survived'] = df_clean['Survived'].astype('category')

print("Shape before:", df.shape)
print("Shape after:", df_clean.shape)
print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())
df_clean.info()
```

```python
df_clean.to_csv("cleaned_data.csv", index=False)
```
